```python
import pandas as pd
import folium

coords = pd.read_csv("iceland_towns_coordinates.csv")

m = folium.Map(location=[65.0, -18.5], zoom_start=6, tiles="CartoDB positron")

for _, row in coords.iterrows():
    folium.CircleMarker(
        location=[row["latitude"], row["longitude"]],
        radius=4,
        color="#185FA5",
        fill=True,
        fill_color="#185FA5",
        fill_opacity=0.9,
        popup=f"<b>{row['town']}</b><br>{row['latitude']:.4f}, {row['longitude']:.4f}",
        tooltip=row["town"],
    ).add_to(m)

m.save("iceland_towns_map.html")
m  # renders inline in Jupyter
```
